```python
df_drop = df.dropna()

print(df_drop)
```

```python
df_fill = df.copy()

df_fill = df_fill.fillna(0)

print(df_fill)
```

```python
df_median = df.copy()

for col in df_median.select_dtypes(include='number').columns:
    df_median[col] = df_median[col].fillna(df_median[col].median())

print(df_median)
```

```python
df_mode = df.copy()

for col in df_mode.columns:
    if df_mode[col].isnull().any():
        df_mode[col] = df_mode[col].fillna(df_mode[col].mode()[0])

print(df_mode)
```

```python
# Forward fill
df_forward = df.ffill()
print(df_forward)

# Backward fill
df_backward = df.bfill()
print(df_backward)
```

```python
df = df.drop_duplicates()

print(df)
```

```python
# Check data types
print(df.dtypes)

# Convert numeric columns into numeric data types
numeric_columns = ['likes', 'comments', 'shares',
                   'watch_time', 'followers']

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

print(df.dtypes)
```

```python
if 'gender' in df.columns:
    df['gender'] = df['gender'].astype('string').str.strip().str.lower()

    df['gender'] = df['gender'].replace({
        'm': 'male',
        'f': 'female',
        'man': 'male',
        'woman': 'female'
    })

    print(df['gender'].value_counts())
```

```python
columns = ['likes', 'comments', 'shares']

for col in columns:
    if col in df.columns:
        df.loc[df[col] < 0, col] = np.nan

print(df[columns].describe())
```

```python
if 'hashtags' in df.columns:
    df['hashtag_count'] = df['hashtags'].fillna('').apply(
        lambda x: len([
            word for word in str(x).split()
            if word.startswith('#')
        ])
    )

print(df[['hashtags', 'hashtag_count']].head())
```

```python
if 'sentiment' in df.columns:
    df['sentiment'] = (
        df['sentiment']
        .astype('string')
        .str.strip()
        .str.lower()
    )

    print(df['sentiment'].value_counts())
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Create simple sample data
np.random.seed(42)
df = pd.DataFrame(
    {
        "impressions": np.random.randint(1000, 50000, 100),
        "likes": np.random.randint(50, 4500, 100),
        "category": np.random.choice(["Tech", "Lifestyle", "Education"], 100),
        "gender": np.random.choice(["Male", "Female"], 100),
        "age": np.random.randint(18, 60, 100),
        "engagement_rate": np.random.uniform(0.01, 0.15, 100),
        "date": pd.date_range(start="2026-01-01", periods=100),
    }
)
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(df["impressions"], df["likes"], color="blue", alpha=0.7)
plt.title("Scatter: Likes vs Impressions")
plt.xlabel("Impressions")
plt.ylabel("Likes")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
daily = df.groupby("date")["engagement_rate"].mean()
plt.plot(daily.index, daily.values, color="orange", linewidth=2)
plt.title("Line: Daily Engagement Trend")
plt.xlabel("Date")
plt.ylabel("Engagement Rate")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
counts = df["category"].value_counts()
plt.bar(counts.index, counts.values, color="green")
plt.title("Bar: Posts by Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(5, 5))
gender_counts = df["gender"].value_counts()
plt.pie(
    gender_counts.values, labels=gender_counts.index, autopct="%1.1f%%"
)
plt.title("Pie: Gender Distribution")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.hist(df["age"], bins=10, color="purple", edgecolor="black")
plt.title("Histogram: Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.boxplot(df["engagement_rate"])
plt.title("Box: Engagement Rate")
plt.ylabel("Engagement Rate")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(
    x="category", y="likes", data=df, estimator=np.mean, palette="muted"
)
plt.title("Bar Plot: Avg Likes by Category")
plt.show()
```
