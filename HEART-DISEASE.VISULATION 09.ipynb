```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("heart-disease.csv")
```

```python
# LINE CHART

import matplotlib.pyplot as plt
import seaborn as sns

# Calculate average cholesterol for each age
line_data = df.groupby('age')['chol'].mean().reset_index()

# Create line chart
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=line_data,
    x='age',
    y='chol',
    marker='o',
    color='blue'
)

plt.title('Age vs Average Cholesterol Level')
plt.xlabel('Age')
plt.ylabel('Average Cholesterol Level')

plt.grid(True)
plt.show()
```

Conclusion:
The line chart helps us understand how average cholesterol levels vary with age. It shows the changes and patterns in cholesterol levels among different age groups. This visualization helps identify possible relationships between age and cholesterol levels in patients.
"The line chart illustrates the relationship between age and average cholesterol levels. It helps identify variations in cholesterol levels across different age groups and provides a clear visual understanding of the dataset."

```python
# BAR CHART

import matplotlib.pyplot as plt
import seaborn as sns

# Calculate average cholesterol by gender
bar_data = df.groupby('sex')['chol'].mean().reset_index()

# Create bar chart
plt.figure(figsize=(8, 5))

sns.barplot(
    data=bar_data,
    x='sex',
    y='chol',
    hue='sex',
    palette='Set2',
    legend=False
)

plt.title('Average Cholesterol Level by Gender')
plt.xlabel('Gender (0 = Female, 1 = Male)')
plt.ylabel('Average Cholesterol Level')

plt.show()
```

CONCLUSION
"The bar chart represents the comparison of average cholesterol levels between male and female patients in the Heart Disease Dataset. It helps identify differences in cholesterol levels between the two gender groups and makes the comparison easy to understand."

```python
# PIE CHART

import matplotlib.pyplot as plt

# Count male and female patients
gender_count = df['sex'].value_counts().sort_index()

# Create pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    gender_count,
    labels=['Female', 'Male'],
    autopct='%1.1f%%',
    colors=['lightpink', 'skyblue'],
    startangle=90,
    explode=(0.05, 0)
)

plt.title('Gender Distribution of Heart Disease Patients')

plt.show()
```

CONCLUSION
The pie chart represents the gender distribution of patients in the Heart Disease Dataset. It shows the percentage of male and female patients using different colours. This visualization helps us understand the proportion of each gender in the dataset

```python
import matplotlib.pyplot as plt
import seaborn as sn
import numpy as np

# Select numerical columns
numeric_cols = df.select_dtypes(include=np.number).columns

# Set figure size
plt.figure(figsize=(18, 10))

# Create box plots
for i, col in enumerate(numeric_cols, 1):
    plt.subplot(4, 4, i)
    sn.boxplot(y=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

box plot visualization helps us understand the distribution of numerical variables in the Heart Disease dataset. It shows the median, quartiles, spread of data, and presence of outliers in different features.

From the box plots, we can observe that:
	1.	Different numerical features have different ranges and distributions.
	2.	Some features contain outliers, indicating unusual or extreme values.
	3.	Variables such as age, cholesterol, resting blood pressure, and maximum heart rate show variations in their distributions.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
sn.scatterplot(data=df, x='age', y='chol', hue='sex', palette='Set1', alpha=0.8)
plt.title('Age vs Cholesterol')
plt.xlabel('Age')
plt.ylabel('Cholesterol')
plt.show()
```

CONCLUSION
Based on the scatter plot generated from the heart disease dataset, here is a concise conclusion summarizing the insights from the Age vs. Cholesterol distribution:
Conclusion: Age vs. Cholesterol Analysis
* Spread & Concentration: The majority of individuals in the dataset fall within the 40 to 70 age range, with cholesterol levels predominantly clustered between 200 mg/dl and 300 mg/

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
sn.violinplot(data=df, x='cp', y='chol', hue='sex', palette='muted', split=True)
plt.title('Cholesterol Distribution by Chest Pain Type (cp) and Sex')
plt.xlabel('Chest Pain Type (cp)')
plt.ylabel('Cholesterol')
plt.show()
```

Conclusion: Cholesterol Distribution by Chest Pain Type and Sex
* Cholesterol Range & Median: Across most chest pain categories (cp), cholesterol values are heavily concentrated between 200 mg/dl and 300 mg/dl, with median levels relatively stable across the different types.
* Distribution Shape: The shape of the violins highlights the data density, showing that while most patients cluster within normal-to-borderline high ranges, extreme high-cholesterol outliers (above 400 mg/dl) appear across various chest pain classifications.
* Demographic Comparison (sex): Utilizing the split=True parameter allows for a direct visual comparison of cholesterol distribution patterns between the two gender categories within each chest pain group, revealing similar spread tendencies despite varying sample concentrations.

```python
#heat map
import matplotlib.pyplot as plt

# Calculate the correlation matrix
corr_matrix = df.corr()

# Create the heatmap using the 'sn' alias for seaborn[span_1](start_span)[span_1](end_span)
plt.figure(figsize=(10, 8))
sn.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Correlation Heatmap - Heart Disease Dataset')
plt.show()
```

  CONCLUSION
  the correlation heatmap helps identify the important factors related to heart disease. Features like chest pain, maximum heart rate, exercise-induced angina, and ST depression show stronger relationships with the target variable. This analysis provides useful insights for understanding heart disease patterns and building predictive machine learning models.

```python
# COUNT PLOT

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.countplot(x='target', data=df, palette='coolwarm')

plt.title('Count Plot - Heart Disease Dataset')
plt.xlabel('Heart Disease (0 = No, 1 = Yes)')
plt.ylabel('Count')

plt.show()
```

Conclusion
The count plot shows the distribution of people with and without heart disease in the dataset. It helps us understand the number of affected and unaffected individuals and provides a clear overview of the dataset.

```python
# HISTOGRAM

import matplotlib.pyplot as plt
import seaborn as sns

df.hist(figsize=(12, 10), bins=20, color='skyblue', edgecolor='black')

plt.suptitle('Histogram - Heart Disease Dataset')
plt.tight_layout()
plt.show()
```

CONCLUSION
 histogram helps visualize the distribution of numerical features in the heart disease dataset. It shows the frequency, spread, and patterns of different attributes, making the data easier to understand and analyze.
